Question 26:- Remove Outliers for volume column.

```python
row = len(df)

Q1 = df["Volume"].quantile(0.25)
Q3 = df["Volume"].quantile(0.75)

IQR = Q3 - Q1
 
lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

df = df[(df["Volume"] >= lower) & (df["Volume"] <= upper)]

print("Rows Before :", row)
print("Rows After  :", len(df))
print("Outliers Removed :", row - len(df))
```

Question 29:-Using sklearn (Multiple linear Regression).

```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error
import numpy as np
```

```python
df["Next_Day_Close"] = df["Close"].shift(-1)
# Remove last row
df.dropna(inplace=True)
df.head()
```

```python
X = df[["Open","High","Low","Close","Volume"]]
y = df["Next_Day_Close"]
```

```python
from sklearn.model_selection import train_test_split

X_train,X_test,y_train,y_test = train_test_split(X,y,test_size=0.2,shuffle=False)
```

```python
X_test
```

```python
from sklearn.linear_model import LinearRegression
model = LinearRegression()
model.fit(X_train,y_train)
```

```python
y_pred = model.predict(X_test)
```

```python
print("Intercept")
print(model.intercept_)
print("Coefficients")
for name,value in zip(X.columns,model.coef_):
    print(name,"=",value)
```

```python
print("MSE :",mean_squared_error(y_test,y_pred))
```

Question 30:- without sklearn.

```python
import pandas as pd
import numpy as np

x1=df[['Open','High','Low','Close','Volume']]

x1=(x1-np.mean(x1,axis=0))/np.std(x1,axis=0)


y1=df['Next_Day_Close'].values.astype(float)
n=len(x1)

m=np.zeros(5)
c=0.0
lr=0.0001
epochs=5000

loss_hist=[]
for i in range(epochs):
    y1_pred=np.dot(x1,m)+c
    err=y1_pred-y1

    loss=(1/n)*np.sum(err**2)
    loss_hist.append(loss)

    dm=(2/n)*np.dot(x1.T,err)
    dc=(2/n)*np.sum(err)

    m=m-lr*dm
    c=c-lr*dc

    if i%500==0:
        print(f'Epoch {i}: Loss={loss:.2f}')
print('Final parameters -> m:', m, ' c:', c)
```

Question 2:-Calculate Rss,MSE,r2_score,Mae.

```python
from sklearn.metrics import mean_squared_error
mse = mean_squared_error(y_test, y_pred)
print("Mean squared error :", mse)
```

```python
from sklearn.metrics import r2_score
r2 = r2_score(y_test, y_pred)
print("R² Score :", r2)
```

```python
import numpy as np
rss = np.sum((y_test - y_pred) ** 2)
print("Residual Sum of Squares (RSS):", rss)
```

```python
from sklearn.metrics import mean_absolute_error

mae = mean_absolute_error(y_test, y_pred)

print("Mean Absolute error :", mae)
```
